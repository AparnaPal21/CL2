```python
import sys
from pathlib import Path
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords

EXAMPLE_TEXT = (
    "The students are learning about text preprocessing. "
    "They were removing common words and stemming the remaining words."
)

def preprocess(text: str) -> str:
    """Lowercase text, remove stop words/punctuation, and stem each word."""
    stemmer = PorterStemmer()
    stop_words = set(stopwords.words("english"))
    # Keep alphabetic tokens only; this also removes punctuation and numbers.
    words = [word.lower() for word in text.split()]
    words = ["".join(char for char in word if char.isalpha()) for word in words]
    words = [word for word in words if word and word not in stop_words]
    return " ".join(stemmer.stem(word) for word in words)

def main() -> None:
    if len(sys.argv) > 1:
        input_path = Path(sys.argv[1])
        text = input_path.read_text(encoding="utf-8")
    else:
        text = EXAMPLE_TEXT
    print("Original text:")
    print(text)
    print("\nPreprocessed text (stop words removed, stemmed):")
    print(preprocess(text))
if __name__ == "__main__":
        main()
```
