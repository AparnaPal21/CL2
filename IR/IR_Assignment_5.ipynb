```python
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

def main() -> None:
    # Load a small, well-known dataset included with scikit-learn.
    iris = load_iris()
    X = iris.data
    # Scale measurements so each feature has a similar influence.
    X_scaled = StandardScaler().fit_transform(X)
    # Merge nearby points into three clusters using Ward's method.
    model = AgglomerativeClustering(n_clusters=3, linkage="ward")
    labels = model.fit_predict(X_scaled)
    print("Dataset: Iris")
    print("Number of samples:", len(X))
    print("Number of clusters:", len(set(labels)))
    print("Cluster sizes:")
    for cluster_id in sorted(set(labels)):
        print(f"  Cluster {cluster_id}: {(labels == cluster_id).sum()} samples")
    # Reduce to two dimensions for a simple plot.
    points_2d = PCA(n_components=2).fit_transform(X_scaled)
    plt.figure(figsize=(8, 5))
    scatter = plt.scatter(
        points_2d[:, 0], points_2d[:, 1], c=labels, cmap="viridis", s=45
    )
    plt.title("Agglomerative clustering of Iris measurements")
    plt.xlabel("Principal component 1")
    plt.ylabel("Principal component 2")
    plt.colorbar(scatter, label="Cluster")
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    main()
```
